```python
import pandas as pd
import psycopg2

# DB 연결
conn = psycopg2.connect(
    host="localhost",
    dbname="land",
    user="postgres",
    password="gkdustn1!",
    port=5433
)
conn.autocommit = False
cur = conn.cursor()
```

```python
try:
    for index, row in df.iterrows():
        # 1️⃣ building_id INSERT
        cur.execute("""
            INSERT INTO building_id (register_id_number , delete_flag)
            VALUES (%s , %s)
            RETURNING bd_number
        """, (0,False))
        bd_number = cur.fetchone()[0]

        # 2️⃣ building_info INSERT
        info_cols = []
        info_vals = []

        for excel_col, db_col in INFO_COLUMN_MAP.items():
            if excel_col in row:
                info_cols.append(db_col)
                info_vals.append(row[excel_col])

        # 0버째에 bd_number을 생성하고, building_id에서 넘겨받은 bd_number를 값으로 넣는다.
        info_cols.insert(0, 'bd_number')
        info_vals.insert(0, bd_number)

        cur.execute(f"""
            INSERT INTO building_info ({','.join(info_cols)})
            VALUES ({','.join(['%s'] * len(info_vals))})
        """, info_vals)

        ############ memo table insert ####################
        # 3️⃣ building_memo INSERT
        memo_cols = []
        memo_vals = []

        for excel_col, db_col in MEMO_COLUMN_MAP.items():
            if excel_col in row:
                memo_cols.append(db_col)
                memo_vals.append(row[excel_col])

        memo_cols.insert(0, 'bd_number')
        memo_vals.insert(0, bd_number)

        cur.execute(f"""
            INSERT INTO building_memo ({','.join(memo_cols)})
            VALUES ({','.join(['%s'] * len(memo_vals))})
        """, memo_vals)

    conn.commit()

except Exception as e:
    conn.rollback()
    print("error")
    raise e

finally:
    cur.close()
    conn.close()
```
